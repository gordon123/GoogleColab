## หน่วยความจำที่มากขึ้น

ผู้ใช้ที่ซื้อแพ็กเกจแบบมีค่าใช้จ่ายอย่างใดอย่างหนึ่งของ Colab จะมีสิทธิ์เข้าถึง VM หน่วยความจำสูงเมื่อมีให้บริการ GPU ที่มีประสิทธิภาพมากขึ้นมาพร้อมกับ VM หน่วยความจำสูงเสมอ
คุณดูปริมาณหน่วยความจำที่เหลืออยู่ได้ทุกเมื่อโดยเรียกใช้เซลล์โค้ดต่อไปนี้ หากผลของการเรียกใช้เซลล์โค้ดด้านล่างคือ "ไม่ได้ใช้รันไทม์แบบ RAM สูง" คุณสามารถเปิดใช้รันไทม์แบบ RAM สูงผ่าน<code>รันไทม์ &gt; เปลี่ยนประเภทรันไทม์</code>ในเมนู จากนั้นเลือก "RAM สูง" ในปุ่มเปิด/ปิด "รูปร่างรันไทม์" แล้วจึงเรียกใช้เซลล์โค้ดอีกครั้ง

```
import psutil

ram_gb = psutil.virtual_memory().total / 1e9
print('Your runtime has {:.1f} gigabytes of available RAM\n'.format(ram_gb))

if ram_gb < 20:
  print('Not using a high-RAM runtime')
else:
  print('You are using a high-RAM runtime!')
```
